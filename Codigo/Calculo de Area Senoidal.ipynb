```python
import math
import numpy as np
from scipy.integrate import quad

c0 = 343.0
d0 = 0.0010  # Diámetro del orificio: 1.0 mm
Ts = 0.0055  # Espesor del septo: 5.5 mm

# Paredes de la celda sepia (medidas de centro a centro)
Tw = 0.0012         # Espesor la pared 1.2 mm
A  = 0.0030         # Amplitud: 3.0 mm
lam = 0.0150        # Longitud de onda (lambda): 15.0 mm
W  = 0.0150         # Separación paredes sinusoidales 15.0 mm
Lx = lam            # Largo de la celda en X (1 ciclo = 15.0 mm)

assert A < 0.5 * (lam - Tw), "Error: Amplitud A muy grande, las paredes chocan."

# Longitud de arco de la pared ondulada
dy_dx = lambda x: A * (2 * math.pi / lam) * math.cos((2 * math.pi / lam) * x)
longitud_arco, _ = quad(lambda x: math.sqrt(1 + dy_dx(x)**2), 0, Lx)

# Dimensiones de las 2 paredes rectas de cierre (en mm)
largo_recta_centro_mm = W * 1000
largo_recta_int_mm    = (W - Tw) * 1000
largo_recta_ext_mm    = (W + Tw) * 1000
espesor_pared_mm      = Tw * 1000

# Área bruta menos área sólida de las 2 paredes onduladas y 2 rectas
area_bruta_m2   = W * Lx
area_paredes_m2 = (longitud_arco * Tw) + ((W - Tw) * Tw)
A_neta_m2       = area_bruta_m2 - area_paredes_m2

l_eff = Ts + 0.85 * d0

#Frecuencias
frecuencias = [400, 500, 600, 700, 800, 900, 1000]

print("--- MEDIDAS DE LAS PAREDES RECTAS DE CIERRE (Fijas para toda la tabla) ---")
print(f"Espesor (Tw)           : {espesor_pared_mm:.2f} mm")
print(f"Largo interno (aire)   : {largo_recta_int_mm:.2f} mm")
print(f"Largo centro a centro  : {largo_recta_centro_mm:.2f} mm")
print(f"Largo exterior total   : {largo_recta_ext_mm:.2f} mm")
print("-" * 82)

for f in frecuencias:
    Vc = (c0**2 * d0**2) / (16.0 * math.pi * l_eff * f**2)
    Vc_cm3 = Vc * 1e6
    
    H_int_mm = (Vc / A_neta_m2) * 1000
    H_ext_mm = H_int_mm + (2 * Ts * 1000)
    
    print(f"f = {f:4d} Hz | Vc = {Vc_cm3:6.2f} cm³ | Alto pared (H_int) = {H_int_mm:6.2f} mm | Alto total bloque = {H_ext_mm:6.2f} mm")

#imprimir ancho de las paredes de cierres planas
print(f"Ancho de las paredes planas: {W * 1000:.2f} mm")
```
