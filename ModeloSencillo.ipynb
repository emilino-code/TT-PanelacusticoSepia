```python
#programa rapido para calcular la frecuencia natural de un resonador de helmoltz
import math

Vc = 0.0005 #500 cm¨3
Ts = 0.05   #5cm
d0 = 0.02    #2cm
c = 343.0    

fn = (c * d0 / 4.0) * math.sqrt(1.0 / (math.pi * Vc * Ts))

print(f"Frecuencia natural: {fn:.2f} Hz")
```

```python
# Barrido de medidas de longitud efectiva vs volumen fijo
Ts_valores = [0.05, 0.075, 0.1, 0.125, 0.15]  #m
Vc_fija = 0.0005  # 500 cm^3
d0 = 0.02         # 2 cm
c = 343.0

fn_valores1 = []
for Ts in Ts_valores:
    fn = (c * d0 / 4.0) * math.sqrt(1.0 / (math.pi * Vc_fija * Ts))
    fn_valores1.append(fn)

print("Frecuencias naturales para diferentes longitudes efectivas:")
for Ts, fn in zip(Ts_valores, fn_valores1):
    print(f"Ts = {Ts:.3f} m: fn = {fn:.2f} Hz")
```

```python
# Ahora Ts, d0 fijos y barrido de volumen
Vc_valores = [0.0001, 0.0002, 0.0003, 0.0004, 0.0005]  # valores de volumen
Ts = 0.05
d0 = 0.02
c = 343.0

fn_valores2 = []
for Vc in Vc_valores:
    fn = (c * d0 / 4.0) * math.sqrt(1.0 / (math.pi * Vc * Ts))
    fn_valores2.append(fn)

print("Frecuencias naturales para diferentes volúmenes:")
for Vc, fn in zip(Vc_valores, fn_valores2):
    print(f"Vc = {Vc} m³: fn = {fn:.2f} Hz")
```

```python
# Ahora Ts fija, Vc fija y d0 variable
d0_valores = [0.01, 0.015, 0.02, 0.025, 0.03]  # diámetro del cuello (m)
Vc_fija = 0.0005  
Ts_fija = 0.05    
c = 343.0         

fn_valores3 = []
for d0 in d0_valores:
    fn = (c * d0 / 4.0) * math.sqrt(1.0 / (math.pi * Vc_fija * Ts_fija))
    fn_valores3.append(fn)

print("Frecuencias naturales para diferentes diámetros de cuello:")
for d0, fn in zip(d0_valores, fn_valores3):
    print(f"d0 = {d0:.3f} m: fn = {fn:.2f} Hz")
```

```python
#graficar cada situación
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(Vc_valores, fn_valores2)
plt.xlabel('Volumen (m^3)')
plt.ylabel('Frecuencia Natural (Hz)')
plt.title('Relación entre Volumen y Frecuencia Natural')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(d0_valores, fn_valores3)
plt.xlabel('Diámetro del Cuello (m)')
plt.ylabel('Frecuencia Natural (Hz)')
plt.title('Relación entre Diámetro del Cuello y Frecuencia Natural')
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(Ts_valores, fn_valores1)
plt.xlabel('Longitud Efectiva (m)')
plt.ylabel('Frecuencia Natural (Hz)')
plt.title('Relación entre Longitud Efectiva y Frecuencia Natural')
plt.show()
```

```python
import math

# Resonador para 400 Hz, 500 Hz, 600 Hz, 700 Hz, 800 Hz, 900 Hz, 1000 Hz
d0 = 0.014   #  2 cm
Ts = 0.03   # 5 cm
c = 343.0  

frecuencias_objetivo = [400, 500, 600, 700, 800, 900, 1000]
volumenes_necesarios = []

for f in frecuencias_objetivo:
    # Despeje riguroso de V a partir de d0:
    V = (c**2 * d0**2) / (16.0 * math.pi * Ts * f**2)
    volumenes_necesarios.append(V)
    print(f"Para f = {f:4d} Hz, V = {V:.6f} m^3 ({V * 1e6:6.2f} cm^3)")
```

```python
#graficamos los volumenes necesarios para cada frecuencia objetivo
plt.figure(figsize=(10, 6))
plt.plot(frecuencias_objetivo, volumenes_necesarios, marker='o')
plt.xlabel('Frecuencia Objetivo (Hz)')
plt.ylabel('Volumen Necesario (m^3)')
plt.grid(True)
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Frecuencias
f = np.linspace(100, 2000, 2000)


f0 = 500       # Frecuencia de resonancia en Hz
amortiguamiento = 0.15

respuesta = 1 / np.sqrt((1 - (f/f0)**2)**2 + (2*amortiguamiento*f/f0)**2)

plt.plot(f, respuesta)
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Respuesta relativa")
plt.title("Respuesta en frecuencia de un resonador")
plt.grid()
plt.show()



f01 = 400
f02 = 800
f03 = 1200

respuesta1 = 1 / np.sqrt((1 - (f/f01)**2)**2 + (2*amortiguamiento*f/f01)**2)

respuesta2 = 1 / np.sqrt((1 - (f/f02)**2)**2 + (2*amortiguamiento*f/f02)**2)

respuesta3 = 1 / np.sqrt((1 - (f/f03)**2)**2 + (2*amortiguamiento*f/f03)**2)

plt.plot(f, respuesta1, label="Resonador 400 Hz")
plt.plot(f, respuesta2, label="Resonador 800 Hz")
plt.plot(f, respuesta3, label="Resonador 1200 Hz")

plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Respuesta relativa")
plt.title("Comparacion de tres resonadores")
plt.legend()
plt.grid()
plt.show()
```
