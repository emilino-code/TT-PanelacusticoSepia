```python
#Codigo Clase San Juditas
import math
import matplotlib.pyplot as plt
import numpy as np

L1 = 1
L2 = 1

inx = 0.5
iny = 0.8

D = math.hypot(inx, iny)

gamma = math.acos((L1**2 + L2**2 - D**2) / (2 * L1 * L2))

alpha = math.acos((L1**2 + D**2 - L2**2) / (2 * L1 * D))

theta = math.atan2(iny, inx) - alpha

theta2 = math.pi - gamma

print("Theta1: ", math.degrees(theta))
print("Theta2: ", math.degrees(theta2))

#plotear los brazos
x1 = L1 * math.cos(theta)
y1 = L1 * math.sin(theta)

x2 = x1 + L2 * math.cos(theta2)
y2 = y1 + L2 * math.sin(theta2)

#plotear los brazos
plt.plot([0, x1], [0, y1])
plt.plot([x1, x2], [y1, y2])
plt.plot(0, 0, 'ro')  
plt.plot(x1, y1, 'ro') 
plt.plot(x2, y2, 'ro')
plt.show()
```
